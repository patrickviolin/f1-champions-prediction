#### 1. Adding the target variable

```python
import numpy as np
import pandas as pd

df = pd.read_csv('../data/02_interim/04_info_after_feature_engineering.csv')
```

#### 2. Dropping info that won't be used in training

```python
df = df.drop(['resultId', 'raceId', 'driverId', 'date', 'statusId', 'status', 'status_group', 'positionOrder', 'laps',
              'qualifyId', 'q1', 'q2', 'q3', 'dob'], axis=1)
```

#### 3. Separate X from y

```python
X = df.drop('is_dnf', axis=1)
y = df[['year', 'is_dnf']]
```

#### 4. Time-series Split - Separate train from test

```python
X_train = X[X['year'] <= 2020]
X_test = X[X['year'] > 2020]

y_train = y[y['year'] <= 2020]
y_test = y[y['year'] > 2020]

X_train = X_train.drop('year', axis=1)
X_test = X_test.drop('year', axis=1)

y_train = y_train.drop('year', axis=1)
y_test = y_test.drop('year', axis=1)
```

#### 5. Export data

```python
datasets = [X_train, X_test, y_train, y_test]
datasets_names = ['X_train', 'X_test', 'y_train', 'y_test']

for i in range(len(datasets)):
    datasets[i].to_csv(f'../data/03_processed/{datasets_names[i]}.csv', index=False)
```
