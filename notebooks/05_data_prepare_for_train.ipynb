#### 1. Load feature-engineered data

```python
import pandas as pd

df = pd.read_csv('../train_and_test_data/02_interim/04_info_after_feature_engineering.csv')
```

#### 2. Separate target from features

```python
df = df.sort_values('raceId').reset_index(drop=True)

y = df[['year', 'positionOrder']]

X = df.drop(['resultId', 'date', 'statusId', 'status', 'status_group', 'positionOrder', 'laps',
             'qualifyId', 'q1', 'q2', 'q3', 'dob', 'is_dnf'], axis=1)
```

#### 3. Review features

```python
X.head()
```

#### 4. Time-series Split - Separate train from test

```python
X_train = X[X['year'] <= 2020]
X_test = X[X['year'] > 2020]

y_train = y[y['year'] <= 2020]
y_test = y[y['year'] > 2020]

# Query ID for XGBoost Ranker
qid_train = X_train[['raceId']].reset_index(drop=True)
qid_test = X_test[['raceId']].reset_index(drop=True)
year_train = X_train[['year']].reset_index(drop=True)
year_test = X_test[['year']].reset_index(drop=True)

driver_train = X_train[['driverId']].reset_index(drop=True)
driver_test = X_test[['driverId']].reset_index(drop=True)

X_train = X_train.drop(['year', 'raceId', 'driverId'], axis=1)
X_test = X_test.drop(['year', 'raceId', 'driverId'], axis=1)

y_train = y_train.drop('year', axis=1)
y_test = y_test.drop('year', axis=1)
```

#### 5. Export data


```python
datasets = [X_train, X_test, y_train, y_test, year_train, year_test, qid_train, qid_test, driver_train, driver_test]
datasets_names = ['X_train', 'X_test', 'y_train', 'y_test', 'year_train', 'year_test', 'qid_train', 'qid_test',
                  'driver_train',
                  'driver_test']

for i in range(len(datasets)):
    datasets[i].to_csv(f'../data/03_processed/{datasets_names[i]}.csv', index=False)
```
