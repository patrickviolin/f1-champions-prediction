### Dataset Corrections

#### 1. Data Typing

```python
import numpy as np
import pandas as pd

df = pd.read_csv('../data/02_interim/results_races_drivers_constructors.csv')
df
```

```python
df_with_nan = df.replace(to_replace='\\N', value=np.nan)

df_with_nan
```

Dropping columns where NaN is >= 90% of its values

```python
df_size = len(df_with_nan)
print(f"Dataset Size: {df_size}")

drop_threshold = df_size * 0.10

df = df_with_nan.dropna(thresh=drop_threshold, axis="columns")
df
```

Fixing Object Column Types

```python
df['date'] = pd.to_datetime(df['date'])

number_y = df['number_y']
number_y.loc[df['number_y'].isnull()] = 0

df['number_y'] = number_y
df['number_y'] = pd.to_numeric(df['number_y'])

df.info()
```

#### Saving updated dataset

```python
df.to_csv('../data/02_interim/results_races_drivers_constructors_without_object.csv', index=False)
```
