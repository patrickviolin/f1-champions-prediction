#### 1. Load feature-engineered data

```python
import pandas as pd

df = pd.read_csv('../data/02_interim/04_info_after_feature_engineering.csv')
```

#### 2. Separate target from features

```python
y = df[['year', 'positionOrder']]

X = df.drop(['resultId', 'raceId', 'driverId', 'date', 'statusId', 'status', 'status_group', 'positionOrder', 'laps',
             'qualifyId', 'q1', 'q2', 'q3', 'dob', 'is_dnf'], axis=1)
```

#### 3. Review features

```python
X.head()
```

#### 4. Time-series Split - Separate train from test

```python
X_train = X[X['year'] <= 2020]
X_test = X[X['year'] > 2020]

y_train = y[y['year'] <= 2020]
y_test = y[y['year'] > 2020]

train_years = X_train[['year']].reset_index(drop=True)

X_train = X_train.drop('year', axis=1)
X_test = X_test.drop('year', axis=1)

y_train = y_train.drop('year', axis=1)
y_test = y_test.drop('year', axis=1)
```

#### 5. Export data


```python
datasets = [X_train, X_test, y_train, y_test, train_years]
datasets_names = ['X_train', 'X_test', 'y_train', 'y_test', 'train_years']

for i in range(len(datasets)):
    datasets[i].to_csv(f'../data/03_processed/{datasets_names[i]}.csv', index=False)
```
